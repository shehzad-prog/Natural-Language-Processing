```python
# Importing Libraries 
import re
from collections import Counter
from PyPDF2 import PdfReader
!pip install nltk

import nltk

nltk.download("stopwords")

from nltk.corpus import stopwords
```

```python
pdf_path = r"D:\Study\Other Fields\Research Papers\Attention is all you need.pdf"
```

```python
reader = PdfReader(pdf_path)

pages = []

for page in reader.pages:
    text = page.extract_text()
    pages.append(text if text else "")

print("Total pages extracted:", len(pages))
```

```python
# Merge all pages into one string
all_text = " ".join(pages)

# Remove punctuation and digits
clean_text = re.sub(r"[^a-zA-Z\s]", "", all_text)

# Convert to lowercase
clean_text = clean_text.lower()

# Split into tokens
tokens = clean_text.split()

print("Total token count:", len(tokens))
```

```python
stopwords = {
    "the", "is", "a", "an", "and",
    "of", "to", "in", "on", "for",
    "with", "this", "that", "it",
    "as", "at", "by", "from",
    "are", "was", "were", "be"
}

# Remove stopwords
filtered_tokens = [
    word for word in tokens
    if word not in stopwords
]

# Count word frequencies
word_frequency = Counter(filtered_tokens)

# Display top 10 words
print("Top 10 most frequent words:\n")

for word, count in word_frequency.most_common(10):
    print(f"{word}: {count}")
```

```python
keyword = input("Enter a keyword to search: ").lower()

found = False

for page_number, page_text in enumerate(pages, start=1):
    
    # Split page into sentences
    sentences = re.split(r'(?<=[.!?])\s+', page_text)
    
    for sentence in sentences:
        if keyword in sentence.lower():
            print(f"\nPage {page_number}:")
            print(sentence.strip())
            found = True

if not found:
    print(f"\nNo sentence containing '{keyword}' was found.")
```

```python
# Question 2

reader = PdfReader(pdf_path)

pages_data = []

for page_number, page in enumerate(reader.pages, start=1):
    
    raw_text = page.extract_text() or ""
    
    # Count words
    word_count = len(raw_text.split())
    
    # Count characters excluding spaces
    char_count = len(raw_text.replace(" ", "").replace("\n", ""))
    
    page_data = {
        "page_number": page_number,
        "raw_text": raw_text,
        "word_count": word_count,
        "char_count": char_count
    }
    
    pages_data.append(page_data)

print("First 2 entries:\n")

for item in pages_data[:2]:
    print(item)
```

```python
# Pages with more than 0 words
non_empty_pages = [
    page for page in pages_data
    if page["word_count"] > 0
]

if non_empty_pages:
    
    # Page with highest word count
    highest_page = max(
        non_empty_pages,
        key=lambda x: x["word_count"]
    )
    
    print("PAGE WITH HIGHEST WORD COUNT")
    print("Page number:", highest_page["page_number"])
    print("Word count:", highest_page["word_count"])
    print("Content:")
    print(highest_page["raw_text"])
    
    print("\n" + "=" * 60 + "\n")
    
    # Page with lowest non-zero word count
    lowest_page = min(
        non_empty_pages,
        key=lambda x: x["word_count"]
    )
    
    print("PAGE WITH LOWEST NON-ZERO WORD COUNT")
    print("Page number:", lowest_page["page_number"])
    print("Word count:", lowest_page["word_count"])
    
    print("\n" + "=" * 60 + "\n")
    
    # Average word count
    average_words = sum(
        page["word_count"] for page in pages_data
    ) / len(pages_data)
    
    print("AVERAGE WORD COUNT:", round(average_words, 2))

else:
    print("All pages contain zero words.")
```

```python
class PhraseNotFoundError(Exception):
    pass


search_phrase = input("Enter a search phrase: ").lower()

matches = []

for page in pages_data:
    
    text = page["raw_text"].lower()
    
    count = text.count(search_phrase)
    
    if count > 0:
        matches.append(
            (page["page_number"], count)
        )

try:
    if not matches:
        raise PhraseNotFoundError(
            f"Phrase '{search_phrase}' was not found in any page."
        )
    
    print("\nSearch Results:")
    
    for page_number, count in matches:
        print(
            f"Page {page_number}: "
            f"{count} occurrence(s)"
        )

except PhraseNotFoundError as e:
    print("\nPhraseNotFoundError:", e)
```

```python
# Question 3

reader = PdfReader(pdf_path)

text = ""

for page in reader.pages:
    text += (page.extract_text() or "") + " "

before_count = len(text)

# Remove URLs
text = re.sub(r"https?://\S+|www\.\S+", "", text)

# Remove digits
text = re.sub(r"\d+", "", text)

# Keep only alphabets, spaces and full stops
text = re.sub(r"[^a-zA-Z\s.]", "", text)

# Collapse multiple spaces
text = re.sub(r"\s+", " ", text).strip()

after_count = len(text)

print("Character count before cleaning:", before_count)
print("Character count after cleaning:", after_count)
```

```python
cleaned_text = text.lower()

sentences = re.split(r"[.!?]", cleaned_text)

print("Sentence count before removing empty strings:", len(sentences))

sentences = [
    sentence.strip()
    for sentence in sentences
    if sentence.strip()
]

print("Sentence count after removing empty strings:", len(sentences))
```

```python
stop_words = set(stopwords.words("english"))

filtered_sentences = []

for sentence in sentences:
    tokens = sentence.split()
    
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]
    
    if len(tokens) >= 5:
        filtered_sentences.append(" ".join(tokens))

print("Sentence count before filtering:", len(sentences))
print("Sentence count after filtering:", len(filtered_sentences))

print("\n3 Sample Sentences:\n")

for sentence in filtered_sentences[:3]:
    print(sentence)
```

```python
output_file = "nlp_ready_output.txt"

with open(output_file, "w", encoding="utf-8") as file:
    for sentence in filtered_sentences:
        file.write(sentence + "\n")

print("File saved:", output_file)
```

```python
with open(output_file, "r", encoding="utf-8") as file:
    saved_sentences = [
        line.strip()
        for line in file
        if line.strip()
    ]

longest_sentence = max(saved_sentences, key=len)
shortest_sentence = min(saved_sentences, key=len)

print("Total lines written:", len(saved_sentences))

print("\nLongest sentence:")
print(longest_sentence)

print("\nShortest sentence:")
print(shortest_sentence)
```

```python
docx_path = r"D:\Study\Semester - 6\Natural Language Processing\LAB\Lab 02 Task.docx"
```

```python
doc = Document(docx_path)

paragraphs_data = []

for paragraph in doc.paragraphs:
    paragraphs_data.append({
        "text": paragraph.text,
        "style": paragraph.style.name
    })

print("Total paragraphs:", len(paragraphs_data))

unique_styles = set(
    item["style"]
    for item in paragraphs_data
)

print("\nUnique styles:")
for style in unique_styles:
    print(style)
```

```python
headings = []
body = []
other_empty = []

for paragraph in paragraphs_data:
    
    style = paragraph["style"]
    text = paragraph["text"]
    
    if style.startswith("Heading"):
        headings.append(paragraph)
        
    elif style == "Normal" or "Body" in style:
        body.append(paragraph)
        
    else:
        other_empty.append(paragraph)

print("Headings:", len(headings))
print("Body:", len(body))
print("Other/Empty:", len(other_empty))
```

```python
qualifying = []

for paragraph in body:
    
    text = paragraph["text"].strip()
    word_count = len(text.split())
    
    if word_count > 5:
        char_count = len(text.replace(" ", ""))
        
        qualifying.append({
            "text": text,
            "word_count": word_count,
            "char_count": char_count
        })

if qualifying:
    
    highest = max(
        qualifying,
        key=lambda x: x["word_count"]
    )
    
    lowest = min(
        qualifying,
        key=lambda x: x["word_count"]
    )
    
    mean_words = sum(
        x["word_count"] for x in qualifying
    ) / len(qualifying)
    
    print("Highest word count:", highest["word_count"])
    print(highest["text"])
    
    print("\nLowest word count above 5:", lowest["word_count"])
    print(lowest["text"])
    
    print("\nMean word count:", round(mean_words, 2))
    
else:
    print("No qualifying body paragraphs.")
```

```python
style_data = {}

for paragraph in paragraphs_data:
    
    style = paragraph["style"]
    
    if style not in style_data:
        style_data[style] = {
            "count": 0,
            "texts": []
        }
    
    style_data[style]["count"] += 1
    style_data[style]["texts"].append(paragraph["text"])

# Show only first 2 texts if more than 2 exist
for style in style_data:
    if len(style_data[style]["texts"]) > 2:
        style_data[style]["texts"] = style_data[style]["texts"][:2]

# Sort by count descending
style_data = dict(
    sorted(
        style_data.items(),
        key=lambda x: x[1]["count"],
        reverse=True
    )
)

pprint(style_data)
```

```python
# Question 6

from docx import Document
import re
from collections import Counter
```

```python
doc_a_path = r"D:\Study\Semester - 6\Natural Language Processing\LAB\Lab 02 Task.docx"
doc_b_path = r"D:\Study\Semester - 6\Natural Language Processing\LAB\Lab 01 Task.docx"
```

```python
doc_a = Document(doc_a_path)
doc_b = Document(doc_b_path)
```

```python
headings_a = [
    p.text.strip()
    for p in doc_a.paragraphs
    if p.style.name.startswith("Heading")
]

headings_b = [
    p.text.strip()
    for p in doc_b.paragraphs
    if p.style.name.startswith("Heading")
]

set_a = set(headings_a)
set_b = set(headings_b)

print("Common headings:")
print(set_a & set_b)

print("\nHeadings only in Document A:")
print(set_a - set_b)

print("\nHeadings only in Document B:")
print(set_b - set_a)
```

```python
body_a = [
    p.text.strip()
    for p in doc_a.paragraphs
    if p.style.name == "Normal"
]

body_b = [
    p.text.strip()
    for p in doc_b.paragraphs
    if p.style.name == "Normal"
]

combined_body = body_a + body_b

print("Count before deduplication:", len(combined_body))

unique_body = list({
    text.lower(): text
    for text in combined_body
    if text.strip()
}.values())

print("Count after deduplication:", len(unique_body))
```

```python
stopwords = {
    "the", "is", "a", "an", "and",
    "of", "to", "in", "on", "for",
    "with", "this", "that", "it",
    "as", "at", "by"
}

counter = Counter()

for paragraph in unique_body:
    
    text = paragraph.lower()
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    
    words = text.split()
    
    words = [
        word for word in words
        if word not in stopwords
    ]
    
    counter.update(words)

print("Top 20 words:")

for word, count in counter.most_common(20):
    print(f"{word}: {count}")
```

```python
all_headings = headings_a + headings_b

average_length = (
    sum(len(p.split()) for p in unique_body) / len(unique_body)
    if unique_body else 0
)

longest_paragraph = (
    max(unique_body, key=lambda p: len(p.split()))
    if unique_body else ""
)

print("Total headings:", len(all_headings))
print("Unique body paragraphs:", len(unique_body))
print("Vocabulary size:", len(counter))
print("Average paragraph length:", round(average_length, 2))

print("\nLongest paragraph:")
print(longest_paragraph)
```

```python
# Question 8

import requests
from collections import defaultdict

todos_url = "https://jsonplaceholder.typicode.com/todos"

response = requests.get(todos_url)

if response.status_code != 200:
    print("Error:", response.status_code)
    
else:
    todos = response.json()
    
    if not isinstance(todos, list):
        print("Error: response is not a list.")
        
    elif len(todos) == 0:
        print("Error: no todo records found.")
        
    else:
        print("Total todo records:", len(todos))
```

```python
user_todos = defaultdict(
    lambda: {
        "total": 0,
        "completed": 0,
        "incomplete": 0
    }
)

for todo in todos:
    
    user_id = todo["userId"]
    
    user_todos[user_id]["total"] += 1
    
    if todo["completed"]:
        user_todos[user_id]["completed"] += 1
    else:
        user_todos[user_id]["incomplete"] += 1

for user_id in sorted(user_todos):
    
    data = user_todos[user_id]
    
    valid = (
        data["completed"] + data["incomplete"]
        == data["total"]
    )
    
    print(user_id, data, "Valid:", valid)
```

```python
completion_rates = {}

for user_id, data in user_todos.items():
    
    rate = (
        data["completed"] /
        data["total"] *
        100
    )
    
    completion_rates[user_id] = round(rate, 1)

highest_user = max(
    completion_rates,
    key=completion_rates.get
)

lowest_user = min(
    completion_rates,
    key=completion_rates.get
)

average_rate = sum(
    completion_rates.values()
) / len(completion_rates)

print(
    "Highest:",
    highest_user,
    completion_rates[highest_user], "%"
)

print(
    "Lowest:",
    lowest_user,
    completion_rates[lowest_user], "%"
)

print(
    "Overall average:",
    round(average_rate, 1), "%"
)
```

```python
users_response = requests.get(users_url)

if users_response.status_code != 200:
    raise ConnectionError(
        f"Status code: {users_response.status_code}"
    )

users = users_response.json()

user_lookup = {
    user["id"]: user["name"]
    for user in users
}

ranking = sorted(
    completion_rates.items(),
    key=lambda x: x[1],
    reverse=True
)

print(
    f"{'Rank':<6}"
    f"{'Name':<25}"
    f"{'Total Tasks':<15}"
    f"{'Completed':<12}"
    f"{'Completion Rate %':<20}"
)

print("-" * 78)

for rank, (user_id, rate) in enumerate(ranking, start=1):
    
    data = user_todos[user_id]
    
    print(
        f"{rank:<6}"
        f"{user_lookup[user_id]:<25}"
        f"{data['total']:<15}"
        f"{data['completed']:<12}"
        f"{rate:<20}"
    )
```

```python
# Question 9

import pandas as pd
import math


users_url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(users_url)

if response.status_code != 200:
    raise ConnectionError(
        f"Status code: {response.status_code}"
    )

users = response.json()
```

```python
def flatten_user(user):
    return {
        "id": user["id"],
        "name": user["name"],
        "username": user["username"],
        "email": user["email"],
        "city": user["address"]["city"],
        "zipcode": user["address"]["zipcode"],
        "lat": user["address"]["geo"]["lat"],
        "lng": user["address"]["geo"]["lng"],
        "company_name": user["company"]["name"],
        "company_catchphrase": user["company"]["catchPhrase"]
    }


flattened_users = [
    flatten_user(user)
    for user in users
]

for user in flattened_users:
    print(user)
```

```python
df = pd.DataFrame(flattened_users)

df["lat"] = df["lat"].astype(float)
df["lng"] = df["lng"].astype(float)

df["distance_from_origin"] = df.apply(
    lambda row: math.sqrt(
        row["lat"] ** 2 +
        row["lng"] ** 2
    ),
    axis=1
)

print(df)
```

```python
furthest_user = df.loc[
    df["distance_from_origin"].idxmax()
]

print("User with greatest distance:")
print(furthest_user)
```

```python
northern = df[df["lat"] > 0]
southern = df[df["lat"] < 0]

print("Northern Hemisphere:")
print(northern["name"].tolist())

print("\nSouthern Hemisphere:")
print(southern["name"].tolist())

north_average = northern["distance_from_origin"].mean()
south_average = southern["distance_from_origin"].mean()

print("\nNorthern average distance:",
      round(north_average, 2))

print("Southern average distance:",
      round(south_average, 2))
```

```python
posts_url = "https://jsonplaceholder.typicode.com/posts"
comments_url = "https://jsonplaceholder.typicode.com/comments"

posts_response = requests.get(posts_url)
comments_response = requests.get(comments_url)

if posts_response.status_code != 200:
    raise ConnectionError(
        f"Posts status: {posts_response.status_code}"
    )

if comments_response.status_code != 200:
    raise ConnectionError(
        f"Comments status: {comments_response.status_code}"
    )

posts = posts_response.json()
comments = comments_response.json()
```

```python
comment_counts = {}

for post in posts:
    
    count = sum(
        comment["postId"] == post["id"]
        for comment in comments
    )
    
    comment_counts[post["id"]] = count


post_results = [
    (
        post["id"],
        post["title"][:40],
        comment_counts[post["id"]]
    )
    for post in posts
]

post_results.sort(
    key=lambda x: x[2],
    reverse=True
)

print("TOP 5 MOST-COMMENTED POSTS")

for post in post_results[:5]:
    print(post)

print("\nBOTTOM 5 LEAST-COMMENTED POSTS")

for post in post_results[-5:]:
    print(post)
```
