Task 1: Regex Fundamentals- Character Classes & Quantifiers

```python
import re

text = """
Hello World! I have 2 cats and 15 books.
Python was created in 1991. Today, I am learning NLP.
The code contains 1234 and 5678 as sample numbers.
"""

# a) Extract all digits from the text
digits = re.findall(r'\d', text)
print("a) Digits:", digits)


# b) Extract all words that start with a capital letter
capital_words = re.findall(r'\b[A-Z][a-z]*\b', text)
print("b) Capitalized words:", capital_words)


# c) Extract all sequences of exactly 4 digits
four_digits = re.findall(r'\b\d{4}\b', text)
print("c) Four-digit sequences:", four_digits)


# d) Extract all words that are between 3 to 6 characters long
words_3_to_6 = re.findall(r'\b[A-Za-z]{3,6}\b', text)
print("d) Words with 3-6 characters:", words_3_to_6)
```

Task 2: Pattern-Based Extraction — Real World Data

```python
import re

text = """
Please contact us at support@gmail.com or info@company.pk.
Our phone numbers are 0300-1234567 and 0312-7654321.
Order IDs are ORD12345 and ORD67890.
The meeting is on 25/09/2026 and the next one is on 05/10/2026.
The discount is 45% and another offer is 12.5%.
"""
```

```python
emails = re.findall(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b', text)

print("a) Emails:", emails)
```

```python
phone_numbers = re.findall(r'\b03\d{2}-\d{7}\b', text)

print("b) Phone numbers:", phone_numbers)
```

```python
dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b', text)

print("c) Dates:", dates)
```

```python
percentages = re.findall(r'\b\d+(?:\.\d+)?%', text)

print("d) Percentages:", percentages)
```

Task 3: Word Tokenization — NLTK vs spaCy

```python
paragraph = """
Ali doesn't like waking up early. It's difficult for him to attend
morning classes because he's usually tired. However, he enjoys studying
data science and doesn't mind working on interesting projects.
"""
```

```python
pip install nltk
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

nltk_tokens = word_tokenize(paragraph)

print("NLTK Tokens:")
print(nltk_tokens)

print("Total NLTK tokens:", len(nltk_tokens))
```

```python
pip install spacy
```

```python
import spacy

nlp = spacy.blank("en")

doc = nlp(paragraph)

spacy_tokens = [token.text for token in doc]

print("spaCy Tokens:")
print(spacy_tokens)

print("Total spaCy tokens:", len(spacy_tokens))
```

# Differences noticed:
# 1. NLTK and spaCy may tokenize contractions such as "doesn't" and "it's"
#    into separate tokens such as "does", "n't" and "It", "'s".
# 2. spaCy tokens are stored as Token objects internally, while NLTK
#    word_tokenize() directly returns a list of strings.

```python
print("Total NLTK tokens:", len(nltk_tokens))
print("Total spaCy tokens:", len(spacy_tokens))
```

Task 4: Token Attributes — Categorizing Tokens

```python
import spacy

nlp = spacy.blank("en")

sentence = "Ali bought 3 books for $50 and emailed the receipt to ali@gmail.com."

doc = nlp(sentence)
```

```python
doc = nlp(sentence)
```

```python
for token in doc:
    print(token.text, "-> is_alpha:", token.is_alpha)
```

```python
for token in doc:
    print(
        token.text,
        "-> like_num:", token.like_num,
        "| like_email:", token.like_email
    )
```

```python
currency_tokens = []

for token in doc:
    if token.is_currency:
        currency_tokens.append(token.text)

print("Currency tokens:", currency_tokens)
```
