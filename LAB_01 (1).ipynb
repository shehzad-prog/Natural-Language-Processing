```python
# Task 06: String Cleaning Methods

# (a)
text = "   MACHINE Learning is AMAZING!!!   "

text = text.strip()
print("After strip():", text)

text = text.lower()
print("After lower():", text)

text = text.replace("!", "")
print("After replace():", text)

# (b)
words = text.split()
print("Tokens:", words)

uppercase_words = [word.upper() for word in words]
print("Uppercase words:", uppercase_words)

joined_text = " | ".join(uppercase_words)
print("Joined text:", joined_text)

# (c)
sentence = "I love Java and Java is popular"

sentence = sentence.replace("Java", "Python")
print("Updated sentence:", sentence)

print("Position of first 'Python':", sentence.find("Python"))

# (d)
def normalize(text):
    return text.strip().lower().replace(",", "").replace(".", "")

print(normalize("  Hello, WORLD. "))
print(normalize("  NLP, is powerful. "))
```

```python
# Task 07: String Validation Methods

# (a)
text = "NLP is the future of AI research"

print("Starts with NLP:", text.startswith("NLP"))
print("Ends with research:", text.endswith("research"))
print("Position of 'future':", text.find("future"))

# (b)
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        label = "Alphabetic"
    elif token.isdigit():
        label = "Numeric"
    else:
        label = "Mixed"

    print(f"{token}: {label}")

# (c)
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for word in words:
    if word.isalpha():
        pure_words.append(word)
    elif word.isdigit():
        pure_numbers.append(word)
    else:
        mixed_tokens.append(word)

print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)

# (d)
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

for token in ["chat", "99", "gpt4", "bert", "2024", "nlp!"]:
    print(token, "→", validate_token(token))
```

```python
# Task 08: Functions and Lambda for NLP Pipelines

# (a)
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

text = "   HELLO Natural Language Processing   "

print("Function:", clean(text))
print("Lambda:", clean_lambda(text))

# (b)
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [token for token in tokens if len(token) > n]

sentence = "I love working with NLP tools in Python daily"

cleaned = clean(sentence)
tokens = tokenize(cleaned)
result = remove_short(tokens, 3)

print("Result:", result)

# (c)
def word_lengths(tokens):
    return {word: len(word) for word in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]

print("Word lengths:", word_lengths(tokens))

# (d)
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

print(nlp_pipeline("Natural language processing is powerful", 3))
print(nlp_pipeline("Deep learning models understand text", 4))
```

```python
# Task 09: map(), filter(), sorted() and Counter

from collections import Counter

# (a)
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lowercase_words = list(map(lambda word: word.lower(), words))
word_lengths = list(map(lambda word: len(word), words))

print("Lowercase:", lowercase_words)
print("Character lengths:", word_lengths)

# (b)
words_long = list(filter(lambda word: len(word) > 3, words))
words_with_t = list(filter(lambda word: "t" in word.lower(), words))

print("Words longer than 3 characters:", words_long)
print("Words containing 't':", words_with_t)

# (c)
print("Alphabetical:", sorted(words))
print("Length ascending:", sorted(words, key=len))
print("Length descending:", sorted(words, key=len, reverse=True))

# (d)
sentence = "nlp nlp python ai nlp python text ai nlp python text"

word_list = sentence.split()
freq = Counter(word_list)

print("Counter:", freq)
print("3 most common:", freq.most_common(3))
print("Frequency of python:", freq["python"])
print("Total number of words:", len(word_list))
print("Unique vocabulary size:", len(freq))
```
